# Baseline experiment

Este notebook ejecuta el baseline del agente y resume las métricas obtenidas sobre el dataset de benchmark.

```python
from __future__ import annotations

import json
from pathlib import Path

from src.evaluation.dataset import load_dataset
from src.evaluation.runner import run_evaluation
from src.models.agent_config import AgentConfig, FlowStep, ToolSpec


dataset = load_dataset("data/benchmark/samples.jsonl")

baseline_config = AgentConfig(
    system_prompt=(
        "You are a careful Python code reviewer. "
        "List logic bugs, security issues, and maintainability concerns."
    ),
    tools=[ToolSpec(name="code_analysis", description="Analyze structure."),],
    flow=[FlowStep(name="review", instructions="Review the code and return issues."),],
    version=0,
)

result = run_evaluation(baseline_config, dataset)

print("Baseline metrics:")
for metric in result.metrics:
    print(f"- {metric.name}: {metric.value:.3f}")

output_path = Path("data/outputs/baseline_results.json")
output_path.parent.mkdir(parents=True, exist_ok=True)
output_path.write_text(
    json.dumps(
        {
            "score": result.score,
            "metrics": [metric.__dict__ for metric in result.metrics],
            "notes": result.notes,
        },
        indent=2,
    ),
    encoding="utf-8",
)

output_path
```
